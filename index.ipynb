# Geometry Course Atlas

Welcome to a visualization-first geometry curriculum built from executable Jupyter notebooks. The atlas is readable first: course folders contain original prose, generated diagrams, computational experiments, artifacts, and checks. Execution is layered on through Colab, JupyterLite-compatible notebooks, and local runtime profiles.

The machine-readable source of truth is [`course-manifest.json`](course-manifest.json), generated from the course indexes and metadata.


## Atlas Manifest Check

Run the cell below when you want to verify that the generated manifest matches the notebook tree. It does not import the full geometry stack; it only checks local metadata and paths.


````python
from pathlib import Path
import json

root = Path.cwd()
manifest_path = root / "course-manifest.json"
manifest = json.loads(manifest_path.read_text(encoding="utf-8"))

missing = []
for course in manifest["courses"]:
    if not (root / course["path"]).exists():
        missing.append(course["path"])
    for notebook in course["notebooks"]:
        if not (root / notebook["path"]).exists():
            missing.append(notebook["path"])

if missing:
    raise RuntimeError("Manifest references missing paths:\n" + "\n".join(missing[:20]))

summary = manifest["summary"]
print(f"Courses: {summary['course_count']}")
print(f"Course notebooks: {summary['notebook_count']}")
print("Runtime profiles:")
for profile, count in sorted(summary["runtime_profile_counts"].items()):
    print(f"  {profile}: {count}")
````

## Launch Options

- **Read:** open any `00-book-index.ipynb` in GitHub, JupyterLab, or a future static portal.
- **Run in Colab:** use each course or notebook `colab_url` in `course-manifest.json`.
- **Run in JupyterLite:** use only entries marked `runtimes.jupyterlite: true`; these are lightweight/browser-compatible candidates.
- **Run locally:** install a profile from `requirements/`, or use the full `uv` environment.

See [`RUNTIME.md`](RUNTIME.md) for setup cells, profile guidance, and local commands.
