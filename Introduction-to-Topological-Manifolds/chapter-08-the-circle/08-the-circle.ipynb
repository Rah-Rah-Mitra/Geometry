## Chapter Goal

This chapter computes the first nontrivial fundamental group in the course: the circle has fundamental group isomorphic to the infinite cyclic group. The computational goal is to make the proof visible. A loop in the circle can be hard to compare directly, because the angle coordinate is periodic. The covering map

$$
\varepsilon: \mathbb{R} \to S^1, \qquad \varepsilon(r)=e^{2\pi i r}
$$

turns that periodic question into a question about ordinary paths in the real line. A lift of a path in the circle is a continuous choice of real angle measured in turns. If the path is a loop, the endpoint of the lift differs from its starting point by an integer. That integer is the winding number. The main theorem follows once we know that lifting is unique after a starting value is chosen and that homotopies lift consistently.

The notebook treats Chapter 8 as a small machine with four linked parts. First, it inspects the exponential quotient map and its evenly covered arcs. Second, it models lifts of paths and homotopies as continuous real-valued angle data. Third, it converts endpoint differences of lifted loops into the winding number and the isomorphism $\pi_1(S^1,1) \cong \mathbb{Z}$. Fourth, it uses the same integer to classify maps $S^1 \to S^1$ by degree and to run a small applied winding-number lab for vector fields on the punctured plane.

The source pages were inspected with `pdftotext` for orientation, terminology, theorem order, and page span only. All prose, examples, code, diagrams, tables, and checks below are original teaching material for this notebook.

## Concept Section 1: The Covering Map and Evenly Covered Arcs

The exponential map $\varepsilon(r)=e^{2\pi i r}$ is a quotient map from the real line onto the circle. Moving by one unit in $\mathbb{R}$ does not change the point of $S^1$, because one unit of $r$ is one full turn. The real line is therefore not just another way to draw the circle. It is an unwrapped model with infinitely many copies of every sufficiently small arc.

The key local fact is that a short open arc $U\subset S^1$ lifts to a disjoint union of open intervals in $\mathbb{R}$. On each interval, $\varepsilon$ is one-to-one and onto $U$, so the inverse branch is continuous. Such an inverse branch is a local section. This is the point where the chapter shifts from intuitive angle language to rigorous topology: there is no continuous global angle on all of $S^1$, but there are continuous local angles on arcs that do not wrap all the way around.

Inspect the diagram below from top to bottom. The top line is part of $\mathbb{R}$ measured in turns. Each highlighted interval is one sheet above the same highlighted arc in the circle. Choosing one sheet is choosing one local branch of angle. Moving to another sheet changes the branch by an integer, and the exponential projection cannot see that integer shift.

## Concept Section 2: Lifts as Continuous Angle Records

A lift of a map into the circle is a map into $\mathbb{R}$ whose exponential projection gives the original map. In practical terms, it is a continuous angle record measured in turns. The same circle point has many real angles above it, so a starting value is part of the data. Once the starting value is fixed and the domain is connected, the chapter's unique lifting property says the rest of the lift is forced.

The Plotly artifact represents the covering map as a helix: the horizontal coordinate is the real lift value $r$, while the circular coordinates are $\cos(2\pi r)$ and $\sin(2\pi r)$. A path lift is a segment along this helix. If the projected path is a loop, the endpoint lies over the same circle point as the start, so the endpoint difference is an integer. If two lifts of the same path start in different sheets, their difference stays equal to that same integer along the entire path. If they start at the same point, the difference is zero everywhere.

This picture also explains why the identity map $S^1\to S^1$ has no global lift. Going once around the circle would force a continuous angle to increase by one turn and also return to its original value at the identified base point. The real line can record the change, but the circle identification prevents one single global angle branch on all of $S^1$.


```python
s = np.linspace(0, 1, 260)
r_grid = np.linspace(-2.35, 3.35, 900)
fig_cover = go.Figure()
fig_cover.add_trace(go.Scatter3d(
    x=r_grid,
    y=np.cos(2 * np.pi * r_grid),
    z=np.sin(2 * np.pi * r_grid),
    mode='lines',
    line={'color': '#495057', 'width': 4},
    name='epsilon(r) as helix',
))

lift_specs = [
    {'turns': -1, 'start': 0.08, 'color': '#6a4c93'},
    {'turns': 0, 'start': 0.08, 'color': '#2a9d8f'},
    {'turns': 1, 'start': 0.08, 'color': '#e76f51'},
    {'turns': 2, 'start': 0.08, 'color': '#f4a261'},
    {'turns': 3, 'start': 0.08, 'color': '#1d3557'},
]
path_lift_rows = []
for spec in lift_specs:
    lift = spec['start'] + spec['turns'] * s
    projected = epsilon(lift)
    fig_cover.add_trace(go.Scatter3d(
        x=lift,
        y=np.real(projected),
        z=np.imag(projected),
        mode='lines',
        line={'color': spec['color'], 'width': 8},
        name=f"lift endpoint change {spec['turns']}",
    ))
    fig_cover.add_trace(go.Scatter3d(
        x=[lift[0], lift[-1]],
        y=[np.real(projected[0]), np.real(projected[-1])],
        z=[np.imag(projected[0]), np.imag(projected[-1])],
        mode='markers',
        marker={'size': 4, 'color': spec['color']},
        showlegend=False,
    ))
    path_lift_rows.append({
        'turns': spec['turns'],
        'start_lift': float(lift[0]),
        'end_lift': float(lift[-1]),
        'endpoint_difference': float(lift[-1] - lift[0]),
        'endpoint_difference_is_integer': bool(abs((lift[-1] - lift[0]) - round(lift[-1] - lift[0])) < 1e-12),
        'max_projection_modulus_error': float(np.max(np.abs(np.abs(projected) - 1))),
    })

fig_cover.update_layout(
    title='Circle covering map: real lifts run along a helix above S^1',
    scene={
        'xaxis_title': 'lift value r in R',
        'yaxis_title': 'Re epsilon(r)',
        'zaxis_title': 'Im epsilon(r)',
        'aspectmode': 'manual',
        'aspectratio': {'x': 2.8, 'y': 1, 'z': 1},
        'camera': {'eye': {'x': 1.85, 'y': 1.65, 'z': 1.05}},
    },
    margin={'l': 0, 'r': 0, 't': 48, 'b': 0},
)
cover_html = save_plotly_html(fig_cover, HTML / 'circle-covering-lift-tracker.html')

reference_lift = 0.08 + 2 * s
translated_lift = reference_lift + 4
same_start_lift = 0.08 + 2 * s
path_checks = {
    'rows': path_lift_rows,
    'all_endpoint_differences_integer': all(row['endpoint_difference_is_integer'] for row in path_lift_rows),
    'max_modulus_error': max(row['max_projection_modulus_error'] for row in path_lift_rows),
    'translated_lift_difference_constant': bool(np.max(np.abs((translated_lift - reference_lift) - 4)) < 1e-12),
    'same_start_unique_difference_zero': bool(np.max(np.abs(same_start_lift - reference_lift)) < 1e-12),
    'identity_map_global_lift_obstruction': 'a global lift would require one value at the base point and also that value plus one after one circuit',
    'passes': True,
}
path_check_path = save_json(path_checks, CHECKS / 'path-lift-checks.json')
show(cover_html, width=860, height=600)
show(path_check_path)
```

## Concept Section 3: Why the Lifting Theorems Work

The lifting theorems in the chapter are not magic facts about trigonometry. They are local-to-global statements. The evenly covered arcs give local sections, which means a lift can be written locally as a branch of angle. For uniqueness, take two lifts of the same map on a connected domain and look at the set where they agree. Near a point of agreement, both lifts land in the same sheet, so local injectivity of $\varepsilon$ forces them to agree nearby. Near a point of disagreement, the two lifts land in distinct sheets, so they stay apart nearby. The agreement set is both open and closed; connectedness makes it all or nothing.

Homotopy lifting adds one more ingredient: time lives in the compact interval $I$. Around a fixed starting point in the parameter space, finitely many evenly covered arcs cover the time track. That finite cover lets us lift the homotopy strip by strip. On overlaps, uniqueness makes the pieces agree, and the gluing lemma turns local lifts into a global lift. If the original homotopy is stationary on part of the domain, the lifted homotopy is stationary there too, because a constant path has a unique constant lift once its start is fixed.

The proof scaffold below is a dependency graph. Read arrows as "is used to prove" rather than as a chronological list of definitions. The lower path is the computation of $\pi_1(S^1)$: path lifting gives endpoint integers, homotopy lifting makes those integers homotopy invariants, and the standard once-around loop supplies the generator.

## Concept Section 4: Winding Number and the Fundamental Group

For a loop $f:I\to S^1$, choose any lift $\widetilde f:I\to\mathbb{R}$. Because $f(0)=f(1)$, both endpoints of the lift lie in the same fiber of $\varepsilon$, and that fiber differs by integer shifts. The winding number is

$$
N(f)=\widetilde f(1)-\widetilde f(0)\in\mathbb{Z}.
$$

This definition does not depend on the chosen lift: another lift differs by an integer constant, so its endpoint difference is the same. It is also unchanged by path homotopy, because a path homotopy lifts to a path homotopy with fixed endpoints when the original homotopy keeps endpoints fixed. The integer therefore belongs to the path class, not merely to one drawing of the loop.

The fundamental group computation is now arithmetic. Let $\omega(s)=e^{2\pi i s}$ be the standard once-around loop. Its lift is $s$, so $N(\omega)=1$. The loop $\alpha_n(s)=e^{2\pi i n s}$ has lift $ns$, so $N(\alpha_n)=n$. Concatenating loops adds endpoint differences. Thus the map from integers to loop classes sends $n$ to $[\omega]^n$, and the inverse sends a loop class to its winding number. These two maps undo each other, giving $\pi_1(S^1,1)\cong\mathbb{Z}$.


```python
def standard_loop(turns, samples=700, base_turn=0.0):
    t = np.linspace(0, 1, samples)
    lift = base_turn + turns * t
    return t, lift, epsilon(lift)


def concatenate_loops(z_first, z_second):
    return np.concatenate([z_first[:-1], z_second])


winding_specs = [-2, -1, 0, 1, 2, 3]
colors = {-2: '#6a4c93', -1: '#4361ee', 0: '#2a9d8f', 1: '#e76f51', 2: '#f4a261', 3: '#1d3557'}
fig, (ax_circle, ax_lift) = plt.subplots(1, 2, figsize=(12, 5.2), constrained_layout=True)
base = np.linspace(0, 1, 600)
x_base, y_base = unit_circle(base)
ax_circle.plot(x_base, y_base, color='#343a40', linewidth=1.4)

winding_rows = []
for n in winding_specs:
    t, lift, z = standard_loop(n)
    x, y = np.real(z), np.imag(z)
    visible_radius = 1 + 0.035 * n
    ax_circle.plot(visible_radius * x, visible_radius * y, color=colors[n], linewidth=2.2, label=f'N={n}')
    ax_circle.scatter([visible_radius * x[-1]], [visible_radius * y[-1]], color=colors[n], s=24)
    ax_lift.plot(t, lift, color=colors[n], linewidth=2.5, label=f'N={n}')
    ax_lift.scatter([0, 1], [lift[0], lift[-1]], color=colors[n], s=24)
    numeric = numeric_winding(z)
    winding_rows.append({
        'loop': f'alpha_{n}',
        'lift_formula': f'{n} * s',
        'start_lift': float(lift[0]),
        'end_lift': float(lift[-1]),
        'endpoint_difference': float(lift[-1] - lift[0]),
        'numeric_winding_from_unwrapped_phase': numeric,
        'rounds_to_integer': int(round(numeric)),
    })

ax_circle.set_aspect('equal')
ax_circle.axis('off')
ax_circle.set_title('Projected loops in S^1, slightly offset for visibility')
ax_circle.legend(loc='upper right', fontsize=8)
ax_lift.axhline(0, color='#adb5bd', linewidth=0.8)
ax_lift.set_xlabel('path parameter s')
ax_lift.set_ylabel('lift in real turns')
ax_lift.set_title('The lift separates loop classes by endpoint height')
ax_lift.grid(alpha=0.25)
ax_lift.legend(loc='upper left', fontsize=8)

winding_png = save_matplotlib(fig, FIGURES / 'winding-number-lifts.png')
plt.close(fig)
winding_table_path = save_csv(winding_rows, TABLES / 'winding-number-table.csv')

pair_checks = []
for n_val in [-2, -1, 0, 1, 2]:
    for m_val in [-1, 0, 2, 3]:
        _, _, z_n = standard_loop(n_val, samples=350)
        _, _, z_m = standard_loop(m_val, samples=350)
        concatenated = concatenate_loops(z_n, z_m)
        total = numeric_winding(concatenated)
        pair_checks.append({'n': n_val, 'm': m_val, 'numeric_N_concat': total, 'expected_sum': n_val + m_val, 'error': float(abs(total - (n_val + m_val)))})

n, m = sp.symbols('n m', integer=True)
winding_checks = {
    'rows': winding_rows,
    'max_integer_error': float(max(abs(row['numeric_winding_from_unwrapped_phase'] - row['endpoint_difference']) for row in winding_rows)),
    'max_concatenation_error': float(max(item['error'] for item in pair_checks)),
    'symbolic_group_law': str(sp.Eq(n + m, m + n)),
    'generator_omega_winding': 1,
    'pi1_statement': 'N([omega]^n) = n and [f] = [omega]^N(f), so pi_1(S^1,1) is infinite cyclic',
}
winding_checks['passes'] = bool(winding_checks['max_integer_error'] < 1e-10 and winding_checks['max_concatenation_error'] < 1e-10)
winding_check_path = save_json(winding_checks, CHECKS / 'winding-number-checks.json')
show(winding_png, width=900)
display(pd.DataFrame(winding_rows))
show(winding_check_path)
```

## Concept Section 5: Degree of Circle Maps

Once $\pi_1(S^1,1)$ has been identified with $\mathbb{Z}$, every continuous map $\varphi:S^1\to S^1$ carries one integer invariant: the winding number of $\varphi\circ\omega$. This is the degree of $\varphi$. If the map preserves the base point, this is exactly the integer by which the induced homomorphism sends the generator of $\pi_1(S^1,1)$. If the map does not preserve the base point, a rotation can move $\varphi(1)$ back to $1$ without changing winding number.

For the power map $p_n(z)=z^n$, the standard generator is sent to $e^{2\pi i n s}$, so the degree is $n$. Conjugation has degree -1 because it reverses angle. The antipodal map has degree 1 because it is only a rotation. A constant map has degree 0. Degree is homotopy invariant and multiplicative under composition: the degree of $\psi\circ\varphi$ is the product of the two degrees.

The artifact below shows the lifted angle of $\varphi\circ\omega$. The vertical change from $s=0$ to $s=1$ is the degree. A map can move every point and still have degree 1, as a rotation does. Conversely, degree not equal to 1 forces a fixed point by the chapter's theorem, while nonzero degree forces surjectivity.


```python
degree_examples = [
    {'map': 'constant map', 'degree': 0, 'lift': lambda t: np.full_like(t, 0.18), 'color': '#2a9d8f'},
    {'map': 'rotation times z', 'degree': 1, 'lift': lambda t: t + 0.18, 'color': '#e76f51'},
    {'map': 'antipodal -z', 'degree': 1, 'lift': lambda t: t + 0.5, 'color': '#f4a261'},
    {'map': 'conjugation z_bar', 'degree': -1, 'lift': lambda t: -t, 'color': '#4361ee'},
    {'map': 'power z^2', 'degree': 2, 'lift': lambda t: 2 * t, 'color': '#1d3557'},
    {'map': 'power z^-2', 'degree': -2, 'lift': lambda t: -2 * t, 'color': '#6a4c93'},
]

t = np.linspace(0, 1, 420)
fig_degree = go.Figure()
degree_rows = []
for example in degree_examples:
    lift = example['lift'](t)
    image = epsilon(lift)
    net_change = float(lift[-1] - lift[0])
    fig_degree.add_trace(go.Scatter(
        x=t,
        y=lift,
        mode='lines',
        line={'color': example['color'], 'width': 3},
        name=f"{example['map']} (deg {example['degree']})",
    ))
    degree_rows.append({
        'map': example['map'],
        'expected_degree': example['degree'],
        'lift_start': float(lift[0]),
        'lift_end': float(lift[-1]),
        'net_lift_change': net_change,
        'numeric_degree': numeric_winding(image),
        'surjective_expected_by_degree_test': bool(example['degree'] != 0),
        'fixed_point_forced_by_degree_test': bool(example['degree'] != 1),
    })
fig_degree.update_layout(
    title='Degree of a circle map as net change of a lifted angle',
    xaxis_title='standard generator parameter s',
    yaxis_title='lift of phi(omega(s)) in turns',
    legend_title='circle map',
    margin={'l': 45, 'r': 20, 't': 48, 'b': 45},
)
degree_html = save_plotly_html(fig_degree, HTML / 'degree-map-gallery.html')
degree_table_path = save_csv(degree_rows, TABLES / 'degree-map-table.csv')

composition_checks = []
for a in range(-3, 4):
    for b in range(-3, 4):
        lhs = a * b
        sym_lhs = sp.expand(sp.Integer(a) * sp.Integer(b))
        composition_checks.append({'outer_degree': a, 'inner_degree': b, 'composition_degree': lhs, 'sympy_value': int(sym_lhs), 'expected_product': a * b})

sample_fixed_tests = []
theta = np.linspace(0, 1, 2001, endpoint=False)
z = epsilon(theta)
map_values = {
    'constant map': epsilon(np.full_like(theta, 0.18)),
    'conjugation z_bar': np.conjugate(z),
    'power z^2': z**2,
    'antipodal -z': -z,
}
for name, values in map_values.items():
    min_distance = float(np.min(np.abs(values - z)))
    sample_fixed_tests.append({'map': name, 'sample_min_abs_phi_z_minus_z': min_distance})

degree_checks = {
    'rows': degree_rows,
    'max_degree_error': float(max(abs(row['numeric_degree'] - row['expected_degree']) for row in degree_rows)),
    'composition_all_products_match': all(item['composition_degree'] == item['expected_product'] == item['sympy_value'] for item in composition_checks),
    'composition_rule': 'deg(p_a o p_b) = a*b for power maps',
    'sample_fixed_point_distances': sample_fixed_tests,
    'nonzero_degree_forces_surjectivity_statement_recorded': True,
    'degree_not_one_forces_fixed_point_statement_recorded': True,
}
degree_checks['passes'] = bool(degree_checks['max_degree_error'] < 1e-10 and degree_checks['composition_all_products_match'])
degree_check_path = save_json(degree_checks, CHECKS / 'degree-map-checks.json')
show(degree_html, width=860, height=500)
display(pd.DataFrame(degree_rows))
show(degree_check_path)
```

## Applied Lab: Winding Number as Vector-Field Index

The problem set points toward a useful application: a vector field on the punctured plane determines a direction map to the circle wherever the vector is nonzero. If a small circle surrounds an isolated zero of the vector field, normalize the vectors along that circle and compute their winding number. The result is the index of the vector field at that zero.

This lab is intentionally local. We do not solve a global theorem about all singularities in a disk. Instead, we sample four vector fields around the unit circle and watch their direction loops. The identity radial field points once around as the base point goes once around, so its index is 1. The negative radial field also has index 1 because it is the radial field followed by a rotation through half a turn. A linear field with negative determinant reverses direction and has index -1. The quadratic field $(x^2-y^2,-2xy)$ doubles angle and reverses orientation, so its index is -2.

This is the same Chapter 8 invariant in a different costume. The circle covering map supplies the integer, and homotopy lifting explains why small perturbations that avoid zeros cannot change it.


```python
def V1(x, y):
    return x, y


def V2(x, y):
    return -x, -y


def V3(x, y):
    return x + y, x - y


def V4(x, y):
    return x**2 - y**2, -2 * x * y


vector_fields = [
    {'name': 'V1(x,y)=(x,y)', 'func': V1, 'expected_index': 1, 'color': '#2a9d8f'},
    {'name': 'V2(x,y)=(-x,-y)', 'func': V2, 'expected_index': 1, 'color': '#e76f51'},
    {'name': 'V3(x,y)=(x+y,x-y)', 'func': V3, 'expected_index': -1, 'color': '#4361ee'},
    {'name': 'V4(x,y)=(x^2-y^2,-2xy)', 'func': V4, 'expected_index': -2, 'color': '#6a4c93'},
]

circle_t = np.linspace(0, 1, 720)
px = np.cos(2 * np.pi * circle_t)
py = np.sin(2 * np.pi * circle_t)
arrow_t = np.linspace(0, 1, 24, endpoint=False)
arrow_x = np.cos(2 * np.pi * arrow_t)
arrow_y = np.sin(2 * np.pi * arrow_t)

fig, axes = plt.subplots(2, 2, figsize=(10, 9), constrained_layout=True)
index_rows = []
for ax, field in zip(axes.flat, vector_fields):
    ux, uy = field['func'](px, py)
    complex_vectors = ux + 1j * uy
    normalized = complex_vectors / np.abs(complex_vectors)
    index = numeric_winding(normalized)
    aux, auy = field['func'](arrow_x, arrow_y)
    lengths = np.sqrt(aux**2 + auy**2)
    aux = aux / lengths
    auy = auy / lengths
    ax.plot(px, py, color='#343a40', linewidth=1.1)
    ax.quiver(arrow_x, arrow_y, aux, auy, color=field['color'], angles='xy', scale_units='xy', scale=7, width=0.006)
    ax.scatter([0], [0], color='black', s=18)
    ax.set_aspect('equal')
    ax.set_xlim(-1.35, 1.35)
    ax.set_ylim(-1.35, 1.35)
    ax.axis('off')
    ax.set_title(f"{field['name']}\ncomputed index {index:.0f}")
    index_rows.append({
        'vector_field': field['name'],
        'expected_index': field['expected_index'],
        'numeric_index': index,
        'rounded_index': int(round(index)),
        'absolute_error': float(abs(index - field['expected_index'])),
    })

vector_png = save_matplotlib(fig, FIGURES / 'vector-field-index-lab.png')
plt.close(fig)
vector_table_path = save_csv(index_rows, TABLES / 'vector-field-index-table.csv')
vector_checks = {
    'rows': index_rows,
    'max_index_error': float(max(row['absolute_error'] for row in index_rows)),
    'all_indices_match_expected': all(row['rounded_index'] == row['expected_index'] for row in index_rows),
    'interpretation': 'index is the winding number of the normalized vector direction along a small circle around the singularity',
}
vector_checks['passes'] = bool(vector_checks['max_index_error'] < 1e-10 and vector_checks['all_indices_match_expected'])
vector_check_path = save_json(vector_checks, CHECKS / 'vector-field-index-checks.json')
show(vector_png, width=760)
display(pd.DataFrame(index_rows))
show(vector_check_path)
```

## Takeaways

The circle is the first place where the fundamental group sees something nontrivial. The reason is not that the circle is geometrically complicated. It is that a global angle coordinate is impossible. The covering map $\varepsilon:\mathbb{R}\to S^1$ solves the local coordinate problem by unwrapping the circle into infinitely many sheets.

Evenly covered arcs are the local mechanism behind lifting. On each sheet above a small arc, $\varepsilon$ has a continuous inverse branch. Connectedness then gives uniqueness: two lifts of the same map that agree once must agree everywhere. Compactness of the time interval and gluing of local branches give homotopy lifting.

A loop in $S^1$ lifts to a path in $\mathbb{R}$ whose endpoints differ by an integer. That integer is the winding number. It is unchanged by path homotopy, it adds under concatenation, and the standard once-around loop has winding number 1. Therefore $\pi_1(S^1,1)$ is infinite cyclic, generated by the class of the standard loop.

The same integer drives degree theory for maps $S^1\to S^1$. Power maps have degree equal to their exponent, conjugation has degree -1, rotations and the antipodal map have degree 1, and constants have degree 0. Degree classifies circle maps up to homotopy, multiplies under composition, forces surjectivity when nonzero, and forces a fixed point when it is not 1.

Winding number is also an applied diagnostic. Normalized vector directions around an isolated zero define a loop in the circle, and its winding number is the local index. This is why Chapter 8 is more than one computation: it introduces the covering-space method and an integer invariant that will reappear in punctured planes, tori, fixed-point arguments, and higher-dimensional degree theory.
