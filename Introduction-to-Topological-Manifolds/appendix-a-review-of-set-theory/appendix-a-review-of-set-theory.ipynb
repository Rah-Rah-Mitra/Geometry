# Appendix A: Review of Set Theory

**Source Span:** John M. Lee, *Introduction to Topological Manifolds*, Second Edition, Appendix A, printed pp. 381-394; assigned PDF pp. 399-412.

**Appendix Goal:** Build the set-theoretic working language used by the topology chapters: how to define sets by membership, combine them, organize elements by relations, treat functions as structured relations, form products and quotient sets, and recognize the countability hypotheses that later appear in bases, covers, sequences, and compactness arguments.

This notebook is concise as a prerequisite review, but it is standalone. The central habit is to translate every statement about sets into a membership test. Once a membership test is explicit, a union, intersection, complement, product, relation, function, quotient, or countability argument can be checked by finite models before it is used abstractly. The finite models here do not prove every infinite theorem, but they expose the shape of the assertions and identify which equalities depend on extra hypotheses such as injectivity, surjectivity, or nonempty choice sets.

The appendix begins with basic set formation and set operations, moves to Cartesian products, relations, equivalence relations, partitions, order relations, and functions, then reviews number systems, cardinality, indexed families, choice, well-ordering, Zorn's lemma, and abstract disjoint unions. Those topics are prerequisites for the rest of the book because topology constantly forms new spaces from old ones: subspaces are specified subsets, product spaces are Cartesian products, quotient spaces come from equivalence relations, disjoint unions keep copies separated, and second-countability is a countability condition on a family of open sets.

## Translation Guide

| Book language | Computational model in this notebook | What to inspect |
| --- | --- | --- |
| A set is determined by its elements | A finite Python `set` and a Boolean membership column | Equality means identical truth values for every element in the universe |
| Union, intersection, difference, complement | Boolean operations on membership predicates | Distributive and De Morgan laws become row-by-row identities |
| Relation on a set | A directed or undirected graph whose edges are ordered pairs | Reflexive, symmetric, transitive, and antisymmetric tests are graph or pair tests |
| Equivalence relation and quotient set | Connected components after generating the smallest equivalence relation | Each element belongs to exactly one class, and classes partition the original set |
| Partial order | A Hasse diagram for a containment order | Incomparable elements explain why maximal need not mean maximum |
| Function `f: X -> Y` | A dictionary with explicit domain and codomain | Fibers, images, preimages, and inverse behavior are separate objects |
| Cartesian product and indexed family | Tuples or functions from an index set | Products multiply choices; projections recover coordinates |
| Abstract disjoint union | Tagged pairs `(element, tag)` | Tags keep equal-looking elements in different copies distinct |
| Countability | An explicit enumeration of a finite window of an infinite pattern | Diagonal order turns `N x N` and reduced fractions into sequences |

## Set Operations as Membership Tests

A set identity is rarely about the names of the sets. It is about what happens to an arbitrary element. For example, the statement `A union (B cap C) = (A union B) cap (A union C)` says that every element of the surrounding universe has the same membership status on the left and on the right. In a finite model we can inspect the entire truth table. In an infinite proof we cannot inspect every element one by one, but the proof uses the same local test: fix an element `x`, translate membership into logic, and simplify.

This is also the point of the warning about specification. A new set is specified inside an already available ambient set. In computation, that is the finite `U` below. The model is not claiming that every set comes with a preferred universe, only that complements and membership tables are meaningful only after the ambient set has been declared.


```python
U = set(range(1, 13))
A = {1, 2, 3, 4, 7, 8}
B = {3, 4, 5, 6, 8, 9}
C = {2, 4, 6, 8, 10, 12}

law_checks = {
    "union_distributes_over_intersection": A | (B & C) == (A | B) & (A | C),
    "intersection_distributes_over_union": A & (B | C) == (A & B) | (A & C),
    "de_morgan_union": U - (A | B) == (U - A) & (U - B),
    "de_morgan_intersection": U - (A & B) == (U - A) | (U - B),
    "difference_as_intersection_with_complement": A - B == A & (U - B),
}

p, q, r = sp.symbols("p q r", boolean=True)
symbolic_checks = {
    "de_morgan_union_symbolic": bool(
        sp.simplify_logic(sp.Xor(sp.Not(sp.Or(p, q)), sp.And(sp.Not(p), sp.Not(q)))) == sp.false
    ),
    "distributive_symbolic": bool(
        sp.simplify_logic(sp.Xor(p | (q & r), (p | q) & (p | r))) == sp.false
    ),
}

membership_rows = []
for x in sorted(U):
    membership_rows.append(
        {
            "element": x,
            "in_A": x in A,
            "in_B": x in B,
            "in_C": x in C,
            "in_A_union_B": x in (A | B),
            "in_A_cap_B": x in (A & B),
            "in_A_minus_B": x in (A - B),
            "in_complement_A": x in (U - A),
        }
    )
membership_table = pd.DataFrame(membership_rows)
set_table_path = save_csv(membership_rows, TABLE_DIR / "set-operations-membership-table.csv")
set_checks_path = save_json(
    {
        "universe_size": len(U),
        "finite_laws": law_checks,
        "symbolic_boolean_laws": symbolic_checks,
        "operation_cardinalities": {
            "A": len(A),
            "B": len(B),
            "C": len(C),
            "A_union_B": len(A | B),
            "A_cap_B": len(A & B),
            "A_minus_B": len(A - B),
        },
    },
    CHECK_DIR / "set-operations-laws.json",
)

positions = {n: ((n - 1) % 4, 2 - ((n - 1) // 4)) for n in sorted(U)}
panels = [
    ("A, B, C membership", None),
    ("A union B", A | B),
    ("A cap B", A & B),
    ("A minus B", A - B),
    ("U minus A", U - A),
    ("U minus (A union B)", U - (A | B)),
]

fig, axes = plt.subplots(2, 3, figsize=(12, 7))
for ax, (title, subset) in zip(axes.flat, panels):
    ax.set_title(title, fontsize=11)
    ax.set_aspect("equal")
    ax.set_xlim(-0.7, 3.7)
    ax.set_ylim(-0.7, 2.7)
    ax.set_xticks([])
    ax.set_yticks([])
    for spine in ax.spines.values():
        spine.set_visible(False)
    for n, (x, y) in positions.items():
        if subset is None:
            labels = [name for name, S0 in [("A", A), ("B", B), ("C", C)] if n in S0]
            ax.scatter([x], [y], s=780, c="#fef3c7" if labels else "#f8fafc", edgecolors="#0f172a" if labels else "#cbd5e1", linewidths=1.6)
            ax.text(x, y + 0.08, str(n), ha="center", va="center", fontsize=10, weight="bold")
            ax.text(x, y - 0.22, "/".join(labels) if labels else "-", ha="center", va="center", fontsize=7)
        else:
            included = n in subset
            ax.scatter([x], [y], s=760, c="#2563eb" if included else "#e5e7eb", edgecolors="#1e3a8a" if included else "#94a3b8", linewidths=1.5)
            ax.text(x, y, str(n), ha="center", va="center", fontsize=10, color="white" if included else "#334155", weight="bold")

fig.suptitle("Membership predicates over a declared universe U", fontsize=14, weight="bold")
fig.tight_layout(rect=[0, 0, 1, 0.95])
set_fig_path = save_matplotlib(fig, FIG_DIR / "set-operations-membership-grid.png")
plt.close(fig)

assert all(law_checks.values())
assert all(symbolic_checks.values())
display_artifact(rel_to_cwd(set_fig_path), width=900)
membership_table
```

## Relations, Equivalence Classes, and Partial Orders

A relation from `X` to `Y` is a subset of `X x Y`; a relation on `X` is a subset of `X x X`. This definition is intentionally broad. A relation might record adjacency, equality, containment, compatibility, or a rule for identifying points. Extra adjectives impose extra tests.

An equivalence relation is reflexive, symmetric, and transitive. The quotient `X/~` is not a mysterious new operation: it is the set of equivalence classes. In a finite generated example, if we begin with a few declared related pairs and then force symmetry, transitivity, and reflexivity, the classes are exactly the connected components of the undirected graph.

A partial order is different. It is reflexive, transitive, and antisymmetric. The containment order on a power set is the model to remember. The diagram below also marks a common mistake: a maximal element of a subset of a partially ordered set need not be a maximum. A maximum is above every element in the subset; a maximal element merely has no strictly larger element within that subset.


```python
X_rel = ["a", "b", "c", "d", "e", "f"]
seed_relation = {("a", "b"), ("b", "c"), ("d", "e")}
G_seed = nx.Graph()
G_seed.add_nodes_from(X_rel)
G_seed.add_edges_from(seed_relation)
classes = sorted([tuple(sorted(c)) for c in nx.connected_components(G_seed)], key=lambda cls: (len(cls), cls))
equiv_relation = {(x, y) for cls in classes for x in cls for y in cls}


def is_reflexive(relation, domain):
    return all((x, x) in relation for x in domain)


def is_symmetric(relation):
    return all((y, x) in relation for x, y in relation)


def is_transitive(relation):
    return all((x, z) in relation for x, y in relation for yy, z in relation if y == yy)


def is_antisymmetric(relation):
    return all(x == y or (y, x) not in relation for x, y in relation)


partition_checks = {
    "covers_X": set().union(*(set(cls) for cls in classes)) == set(X_rel),
    "all_nonempty": all(len(cls) > 0 for cls in classes),
    "pairwise_disjoint": sum(len(cls) for cls in classes) == len(set().union(*(set(cls) for cls in classes))),
}
equiv_checks = {
    "reflexive": is_reflexive(equiv_relation, X_rel),
    "symmetric": is_symmetric(equiv_relation),
    "transitive": is_transitive(equiv_relation),
    "partition_checks": partition_checks,
    "classes": [list(cls) for cls in classes],
}

base = ("p", "q", "r")
power_nodes = [frozenset(s) for r_size in range(4) for s in itertools.combinations(base, r_size)]
subset_order = {(S, T) for S in power_nodes for T in power_nodes if S.issubset(T)}
family = {frozenset({"p", "q"}), frozenset({"p", "r"})}
maximal_in_family = [S for S in family if not any(S < T for T in family)]
maximum_candidates = [S for S in family if all(T.issubset(S) for T in family)]
poset_checks = {
    "subset_order_reflexive": is_reflexive(subset_order, power_nodes),
    "subset_order_antisymmetric": is_antisymmetric(subset_order),
    "subset_order_transitive": is_transitive(subset_order),
    "family_has_two_maximal_elements": len(maximal_in_family) == 2,
    "family_has_no_maximum": len(maximum_candidates) == 0,
}
relation_checks_path = save_json({"equivalence": equiv_checks, "poset": poset_checks}, CHECK_DIR / "relation-quotient-poset-checks.json")

fig, axes = plt.subplots(1, 2, figsize=(13, 5.7))
class_lookup = {x: idx for idx, cls in enumerate(classes) for x in cls}
class_colors = ["#60a5fa", "#f59e0b", "#34d399", "#f472b6"]
pos_rel = nx.spring_layout(G_seed, seed=11)
nx.draw_networkx_edges(G_seed, pos_rel, ax=axes[0], width=2.2, edge_color="#64748b")
nx.draw_networkx_nodes(G_seed, pos_rel, ax=axes[0], node_size=1100, node_color=[class_colors[class_lookup[n] % len(class_colors)] for n in G_seed.nodes], edgecolors="#0f172a")
nx.draw_networkx_labels(G_seed, pos_rel, ax=axes[0], font_size=11, font_weight="bold")
axes[0].set_title("Generated equivalence classes as components", fontsize=12, weight="bold")
axes[0].text(0.02, -0.08, "Quotient classes: " + ", ".join("{" + ",".join(cls) + "}" for cls in classes), transform=axes[0].transAxes, fontsize=9)
axes[0].axis("off")

H = nx.DiGraph()
H.add_nodes_from(power_nodes)
for S in power_nodes:
    for T in power_nodes:
        if len(T) == len(S) + 1 and S < T:
            H.add_edge(S, T)


def subset_label(S):
    return "{}" if not S else "{" + ",".join(sorted(S)) + "}"


pos_power = {}
for r_size in range(4):
    level = [S for S in power_nodes if len(S) == r_size]
    for j, S in enumerate(sorted(level, key=subset_label)):
        pos_power[S] = (j - (len(level) - 1) / 2, r_size)
node_colors = ["#f97316" if node in family else "#a7f3d0" if node == frozenset(base) else "#e2e8f0" for node in H.nodes]
nx.draw_networkx_edges(H, pos_power, ax=axes[1], arrows=False, width=1.8, edge_color="#94a3b8")
nx.draw_networkx_nodes(H, pos_power, ax=axes[1], node_size=1250, node_color=node_colors, edgecolors="#334155")
nx.draw_networkx_labels(H, pos_power, labels={S: subset_label(S) for S in H.nodes}, ax=axes[1], font_size=8)
axes[1].set_title("Containment poset: maximal need not be maximum", fontsize=12, weight="bold")
axes[1].text(0.02, -0.08, "Orange family has two maximal elements and no single member above both.", transform=axes[1].transAxes, fontsize=9)
axes[1].axis("off")

fig.tight_layout()
relations_fig_path = save_matplotlib(fig, FIG_DIR / "relations-partitions-posets.png")
plt.close(fig)

assert all(equiv_checks[key] for key in ["reflexive", "symmetric", "transitive"])
assert all(partition_checks.values())
assert all(poset_checks.values())
display_artifact(rel_to_cwd(relations_fig_path), width=950)
{"equivalence_classes": classes, "maximal_family": [subset_label(s) for s in maximal_in_family]}
```

## Functions, Images, Preimages, and Inverses

A function is a relation with exactly one output in the codomain for every input in the domain. The domain and codomain are part of the data. This matters: the same ordered pairs can describe a function into a small codomain or into a larger codomain, but surjectivity changes when the codomain changes.

The safest distinction is between images and preimages. Images push subsets of the domain forward to the codomain. Preimages pull subsets of the codomain back to the domain. Preimages behave especially well with Boolean set operations because membership in `f^{-1}(T)` is simply the predicate `f(x) in T`. Images preserve unions, but they do not generally preserve intersections or differences without extra hypotheses. The artifact below displays a surjective, non-injective function. Shared fibers are the visible reason why inverse-function notation and preimage notation should not be confused.


```python
X_fun = list(range(8))
Y_fun = ["alpha", "beta", "gamma", "delta"]
f = {0: "alpha", 1: "alpha", 2: "beta", 3: "gamma", 4: "beta", 5: "delta", 6: "gamma", 7: "delta"}
S = {0, 2, 4, 6}
S_prime = {1, 3, 5}
T = {"beta", "delta"}
T_prime = {"gamma", "delta"}
S_bad = {0, 2}
S_bad_prime = {1, 3}

image = lambda subset: {f[x] for x in subset}
preimage = lambda subset: {x for x in X_fun if f[x] in subset}
fibers = {y: sorted(preimage({y})) for y in Y_fun}
is_injective = len(set(f.values())) == len(X_fun)
is_surjective = set(f.values()) == set(Y_fun)
right_inverse = {y: fibers[y][0] for y in Y_fun if fibers[y]}
right_inverse_valid = all(f[right_inverse[y]] == y for y in Y_fun)
left_inverse_exists = is_injective and len(X_fun) > 0

function_identity_checks = {
    "preimage_preserves_union": preimage(T | T_prime) == preimage(T) | preimage(T_prime),
    "preimage_preserves_intersection": preimage(T & T_prime) == preimage(T) & preimage(T_prime),
    "preimage_preserves_difference": preimage(T - T_prime) == preimage(T) - preimage(T_prime),
    "S_subset_preimage_image_S": S.issubset(preimage(image(S))),
    "image_preserves_union": image(S | S_prime) == image(S) | image(S_prime),
    "image_intersection_subset": image(S_bad & S_bad_prime).issubset(image(S_bad) & image(S_bad_prime)),
    "image_intersection_equality_fails_in_example": image(S_bad & S_bad_prime) != image(S_bad) & image(S_bad_prime),
    "surjective": is_surjective,
    "not_injective": not is_injective,
    "right_inverse_valid": right_inverse_valid,
    "left_inverse_exists": left_inverse_exists,
}

function_rows = []
for x in X_fun:
    function_rows.append({"x": x, "f_x": f[x], "x_in_S": x in S, "f_x_in_T": f[x] in T, "x_in_preimage_T": x in preimage(T)})
function_table_path = save_csv(function_rows, TABLE_DIR / "function-image-preimage-table.csv")
function_checks_path = save_json(
    {
        "domain_size": len(X_fun),
        "codomain_size": len(Y_fun),
        "fibers": fibers,
        "right_inverse": right_inverse,
        "identity_checks": function_identity_checks,
        "counterexample": {
            "S": sorted(S_bad),
            "S_prime": sorted(S_bad_prime),
            "image_intersection_left": sorted(image(S_bad & S_bad_prime)),
            "image_intersection_right": sorted(image(S_bad) & image(S_bad_prime)),
        },
    },
    CHECK_DIR / "function-image-preimage-checks.json",
)

fig, ax = plt.subplots(figsize=(10.5, 6))
ax.set_xlim(-0.2, 1.25)
ax.set_ylim(-0.8, max(len(X_fun), len(Y_fun)) - 0.2)
ax.axis("off")
ax.set_title("A surjective non-injective function: fibers, image, and preimage", fontsize=13, weight="bold")
x_positions = {x: (0, len(X_fun) - 1 - i) for i, x in enumerate(X_fun)}
y_positions = {y: (1, len(Y_fun) - 1 - i * 1.8) for i, y in enumerate(Y_fun)}
for x, (px, py) in x_positions.items():
    in_pre = x in preimage(T)
    ax.scatter([px], [py], s=520, c="#dbeafe" if in_pre else "#f8fafc", edgecolors="#1d4ed8", linewidths=1.4)
    ax.text(px, py, str(x), ha="center", va="center", fontsize=10, weight="bold")
    ax.text(px - 0.11, py, "S" if x in S else "", ha="right", va="center", fontsize=8, color="#7c2d12")
for y, (px, py) in y_positions.items():
    in_T = y in T
    ax.scatter([px], [py], s=760, c="#fed7aa" if in_T else "#f1f5f9", edgecolors="#9a3412", linewidths=1.4)
    ax.text(px, py, y, ha="center", va="center", fontsize=9, weight="bold")
    ax.text(px + 0.12, py, "T" if in_T else "", ha="left", va="center", fontsize=8, color="#7c2d12")
for x, y in f.items():
    start, end = x_positions[x], y_positions[y]
    ax.add_patch(FancyArrowPatch((start[0] + 0.05, start[1]), (end[0] - 0.1, end[1]), arrowstyle="-|>", mutation_scale=11, linewidth=1.2, color="#475569", alpha=0.82, connectionstyle="arc3,rad=0.05"))
ax.text(0, -0.45, "domain X", ha="center", fontsize=10)
ax.text(1, -0.45, "codomain Y", ha="center", fontsize=10)
ax.text(0.48, -0.72, "Blue domain points are f^{-1}(T); orange codomain points are T. Shared arrows form nontrivial fibers.", ha="center", fontsize=9)
function_fig_path = save_matplotlib(fig, FIG_DIR / "functions-image-preimage-fibers.png")
plt.close(fig)

assert all(function_identity_checks[k] for k in function_identity_checks if k != "left_inverse_exists")
assert function_identity_checks["left_inverse_exists"] is False
display_artifact(rel_to_cwd(function_fig_path), width=850)
pd.DataFrame(function_rows)
```

## Products, Indexed Families, Disjoint Unions, and Quotients

Ordered pairs and Cartesian products are the formal mechanism behind coordinates. A point of `X x Y` is not just a two-element set; it remembers first and second coordinates. For a finite product, the projection maps recover those coordinates. For an arbitrary indexed product, the same idea is phrased as a function from the index set: choose one element from each indexed set.

Disjoint unions solve a different problem. If two sets share elements but we want to treat them as separate copies, tags make the copies disjoint. The element called `point` in the red copy is not the element called `point` in the blue copy after tagging.

Quotients go in the opposite direction: instead of separating copies, they deliberately identify elements. A quotient map sends each element to the equivalence class containing it. Later quotient topologies add open-set structure to this set-level operation, but the set-level operation already requires a partition into classes.


```python
A_prod = ["u", "v", "w"]
B_prod = [0, 1]
cartesian_product = list(itertools.product(A_prod, B_prod))
projection_1 = {pair: pair[0] for pair in cartesian_product}
projection_2 = {pair: pair[1] for pair in cartesian_product}
indexed_family = {"red": {"point", "edge"}, "blue": {"point", "face"}, "green": {"edge"}}
disjoint_union = [(value, tag) for tag, values in indexed_family.items() for value in sorted(values)]
untagged_values = [value for value, _tag in disjoint_union]
Q_domain = list(range(10))
quotient_classes = {f"class {r}": [x for x in Q_domain if x % 3 == r] for r in range(3)}
quotient_map = {x: f"class {x % 3}" for x in Q_domain}
product_quotient_checks = {
    "product_cardinality_multiplies": len(cartesian_product) == len(A_prod) * len(B_prod),
    "projection_1_values_in_A": set(projection_1.values()).issubset(set(A_prod)),
    "projection_2_values_in_B": set(projection_2.values()).issubset(set(B_prod)),
    "tagged_disjoint_union_size_is_sum": len(disjoint_union) == sum(len(values) for values in indexed_family.values()),
    "untagged_values_have_repetitions": len(untagged_values) > len(set(untagged_values)),
    "quotient_classes_partition_domain": sorted(sum(quotient_classes.values(), [])) == Q_domain,
    "quotient_map_constant_on_classes": all(quotient_map[x] == label for label, xs in quotient_classes.items() for x in xs),
}
product_rows = []
for pair in cartesian_product:
    product_rows.append({"construction": "product", "object": str(pair), "first_projection": pair[0], "second_projection": pair[1]})
for value, tag in disjoint_union:
    product_rows.append({"construction": "disjoint_union", "object": f"({value}, {tag})", "first_projection": value, "second_projection": tag})
for x, cls in quotient_map.items():
    product_rows.append({"construction": "quotient_map", "object": str(x), "first_projection": cls, "second_projection": x % 3})
product_table_path = save_csv(product_rows, TABLE_DIR / "product-disjoint-union-quotient-data.csv")
product_checks_path = save_json(
    {"cartesian_product": [list(pair) for pair in cartesian_product], "disjoint_union": [list(item) for item in disjoint_union], "quotient_classes": quotient_classes, "checks": product_quotient_checks},
    CHECK_DIR / "product-quotient-invariants.json",
)

fig = make_subplots(rows=1, cols=3, subplot_titles=("Cartesian product", "Tagged disjoint union", "Quotient map"), horizontal_spacing=0.12)
fig.add_trace(go.Scatter(x=[A_prod.index(a) for a, b in cartesian_product], y=[b for a, b in cartesian_product], mode="markers+text", marker=dict(size=22, color="#2563eb"), text=[f"({a},{b})" for a, b in cartesian_product], textposition="top center", hovertext=[f"pi1={a}<br>pi2={b}" for a, b in cartesian_product], hoverinfo="text", showlegend=False), row=1, col=1)
for tag_i, (tag, values) in enumerate(indexed_family.items()):
    sorted_values = sorted(values)
    fig.add_trace(go.Scatter(x=[tag_i] * len(sorted_values), y=list(range(len(sorted_values))), mode="markers+text", marker=dict(size=24, color=["#ef4444", "#3b82f6", "#22c55e"][tag_i]), text=[f"({value},{tag})" for value in sorted_values], textposition="middle right", hovertext=[f"tag={tag}<br>value={value}" for value in sorted_values], hoverinfo="text", showlegend=False), row=1, col=2)
for x in Q_domain:
    cls = quotient_map[x]
    class_index = int(cls.split()[-1])
    fig.add_trace(go.Scatter(x=[0, 1], y=[x, class_index * 3 + 1], mode="lines", line=dict(color="#94a3b8", width=1), hoverinfo="skip", showlegend=False), row=1, col=3)
fig.add_trace(go.Scatter(x=[0] * len(Q_domain), y=Q_domain, mode="markers+text", marker=dict(size=16, color="#0f172a"), text=[str(x) for x in Q_domain], textposition="middle left", hovertext=[f"element {x}" for x in Q_domain], hoverinfo="text", showlegend=False), row=1, col=3)
fig.add_trace(go.Scatter(x=[1, 1, 1], y=[1, 4, 7], mode="markers+text", marker=dict(size=32, color=["#f97316", "#a855f7", "#14b8a6"]), text=["[0]", "[1]", "[2]"], textposition="middle right", hovertext=[f"{label}: {xs}" for label, xs in quotient_classes.items()], hoverinfo="text", showlegend=False), row=1, col=3)
fig.update_xaxes(title_text="first coordinate", tickmode="array", tickvals=list(range(len(A_prod))), ticktext=A_prod, row=1, col=1)
fig.update_yaxes(title_text="second coordinate", tickmode="array", tickvals=B_prod, row=1, col=1)
fig.update_xaxes(title_text="tag", tickmode="array", tickvals=list(range(len(indexed_family))), ticktext=list(indexed_family), row=1, col=2)
fig.update_yaxes(visible=False, row=1, col=2)
fig.update_xaxes(visible=False, row=1, col=3)
fig.update_yaxes(visible=False, row=1, col=3)
fig.update_layout(height=500, width=1050, title_text="Products separate coordinates, tags separate copies, quotients identify elements")
products_html_path = save_plotly_html(fig, HTML_DIR / "products-disjoint-unions-quotient-map.html")

assert all(product_quotient_checks.values())
display_artifact(rel_to_cwd(products_html_path), width=980, height=520)
{"product_size": len(cartesian_product), "disjoint_union_size": len(disjoint_union), "quotient_class_count": len(quotient_classes)}
```

## Countability as an Explicit Listing

Two sets have the same cardinality when a bijection exists between them. For finite sets this is just counting. For infinite sets it becomes a constructive question: can we list the elements without missing any and without assigning two ranks to the same element?

The countability of `N x N` is the prototype. The diagonal listing orders pairs by the sum of their coordinates. Every pair eventually appears because it has a finite coordinate sum. Rational numbers are handled by a related idea: list integer numerator and positive denominator pairs, then keep the reduced fractions while skipping duplicates. This notebook shows only a finite window, but the finite window has the same local checks used in the proof: ranks are unique, reduced fractions are normalized, and the visible part of each diagonal is exhausted before the next diagonal is completed.

These countability facts become topological tools later. A countable basis is an indexed family whose indices can be arranged as a sequence. Countable unions appear in second-countability arguments, and sequences are functions whose domain is `N`.


```python
grid_limit = 8
pairs = [(i, j) for diagonal in range(2, 2 * grid_limit + 1) for i in range(1, grid_limit + 1) for j in [diagonal - i] if 1 <= j <= grid_limit]
pair_rank = {pair: rank + 1 for rank, pair in enumerate(pairs)}
fraction_candidates = []
for height in range(1, 11):
    for q_den in range(1, height + 1):
        for p_num in range(-(height - q_den), height - q_den + 1):
            if math.gcd(abs(p_num), q_den) == 1:
                fraction_candidates.append((p_num, q_den))
seen_fraction_values = set()
reduced_fractions = []
for p_num, q_den in fraction_candidates:
    if (p_num, q_den) not in seen_fraction_values:
        seen_fraction_values.add((p_num, q_den))
        reduced_fractions.append((p_num, q_den))
reduced_fractions = reduced_fractions[:45]
countability_checks = {
    "pair_count_in_window": len(pairs),
    "pair_ranks_unique": len(set(pair_rank.values())) == len(pair_rank),
    "pair_window_covered": set(pair_rank) == {(i, j) for i in range(1, grid_limit + 1) for j in range(1, grid_limit + 1)},
    "fractions_reduced": all(math.gcd(abs(p_num), q_den) == 1 and q_den > 0 for p_num, q_den in reduced_fractions),
    "fraction_representatives_unique": len(reduced_fractions) == len(set(reduced_fractions)),
    "integer_samples_present": all((n, 1) in reduced_fractions for n in range(-3, 4)),
}
countability_checks_path = save_json(
    {
        "grid_limit": grid_limit,
        "first_pairs": [{"rank": pair_rank[pair], "pair": list(pair)} for pair in pairs[:24]],
        "first_reduced_fractions": [{"rank": i + 1, "numerator": p_num, "denominator": q_den, "value": p_num / q_den} for i, (p_num, q_den) in enumerate(reduced_fractions[:30])],
        "checks": countability_checks,
    },
    CHECK_DIR / "countability-enumeration-checks.json",
)

fig = make_subplots(rows=1, cols=2, subplot_titles=("Diagonal enumeration of N x N", "Reduced rational window"), horizontal_spacing=0.15)
fig.add_trace(go.Scatter(x=[i for i, j in pairs], y=[j for i, j in pairs], mode="markers+text", marker=dict(size=18, color=[pair_rank[(i, j)] for i, j in pairs], colorscale="Viridis", showscale=True, colorbar=dict(title="rank")), text=[str(pair_rank[(i, j)]) for i, j in pairs], textposition="middle center", hovertext=[f"rank {pair_rank[(i, j)]}: ({i}, {j})<br>diagonal={i + j}" for i, j in pairs], hoverinfo="text", showlegend=False), row=1, col=1)
fig.add_trace(go.Scatter(x=[p_num / q_den for p_num, q_den in reduced_fractions], y=[0] * len(reduced_fractions), mode="markers+text", marker=dict(size=16, color=list(range(1, len(reduced_fractions) + 1)), colorscale="Plasma", showscale=True, colorbar=dict(title="rank", x=1.02)), text=[str(i + 1) for i in range(len(reduced_fractions))], textposition="top center", hovertext=[f"rank {i + 1}: {p_num}/{q_den} = {p_num / q_den:.3f}" for i, (p_num, q_den) in enumerate(reduced_fractions)], hoverinfo="text", showlegend=False), row=1, col=2)
fig.add_hline(y=0, line_color="#94a3b8", row=1, col=2)
fig.update_xaxes(title_text="i", dtick=1, row=1, col=1)
fig.update_yaxes(title_text="j", dtick=1, row=1, col=1)
fig.update_xaxes(title_text="rational value", range=[-5.5, 5.5], row=1, col=2)
fig.update_yaxes(visible=False, row=1, col=2)
fig.update_layout(height=520, width=1050, title_text="Countability proofs build explicit enumerations")
countability_html_path = save_plotly_html(fig, HTML_DIR / "countability-diagonal-enumeration.html")

assert all(countability_checks.values())
display_artifact(rel_to_cwd(countability_html_path), width=980, height=540)
{"visible_pair_ranks": len(pair_rank), "visible_fraction_ranks": len(reduced_fractions)}
```

## Proof and Invariant Scaffold

The finite checks above are not replacements for proofs, but they are useful proof scaffolds. They isolate the invariant that a proof must preserve.

For set identities, the invariant is membership truth value for an arbitrary element. For equivalence relations, the invariant is that each element has exactly one class and that the relation is reflexive, symmetric, and transitive. For quotient maps, the invariant is constancy on classes: a rule defined on the original set descends to the quotient exactly when equal-class elements receive equal outputs. For functions, the invariant behind a right inverse is surjectivity: every codomain point has at least one selected input. The axiom of choice is the abstract assertion that such selections can be made for arbitrary indexed families of nonempty sets, even when there is no explicit formula for the selected element.

The applied lab below uses only finite data. It models a situation that appears repeatedly in topology: start with a set `X`, impose an equivalence relation, form the quotient `X/~`, and ask whether another map out of `X` really depends only on the quotient class.

## Applied Lab: Does a Rule Factor Through the Quotient?

Given a quotient map `q: X -> X/~`, a map `h: X -> Z` factors through the quotient if there is a function `H: X/~ -> Z` such that `h = H o q`. Set-theoretically, this is exactly the statement that `h` is constant on every equivalence class. If two equivalent elements produce different `h` values, then no well-defined `H` can exist, because the same quotient class would have to receive two different outputs.


```python
lab_X = list(range(9))
lab_classes = {"even-small": [0, 2, 4], "odd-small": [1, 3], "large": [5, 6, 7, 8]}
lab_q = {x: label for label, xs in lab_classes.items() for x in xs}
h_good = {x: len(lab_q[x]) for x in lab_X}
h_bad = {x: x % 4 for x in lab_X}


def factors_through(classes, h):
    witness = {}
    failures = []
    for label, xs in classes.items():
        values = {h[x] for x in xs}
        if len(values) == 1:
            witness[label] = next(iter(values))
        else:
            failures.append({"class": label, "values": sorted(values), "elements": xs})
    return len(failures) == 0, witness, failures


good_factors, good_witness, good_failures = factors_through(lab_classes, h_good)
bad_factors, bad_witness, bad_failures = factors_through(lab_classes, h_bad)
lab_checks = {
    "quotient_partition_covers_X": sorted(sum(lab_classes.values(), [])) == lab_X,
    "quotient_partition_pairwise_disjoint": len(sum(lab_classes.values(), [])) == len(set(sum(lab_classes.values(), []))),
    "good_rule_factors": good_factors,
    "bad_rule_does_not_factor": not bad_factors,
    "bad_failure_recorded": len(bad_failures) > 0,
}
lab_checks_path = save_json(
    {"classes": lab_classes, "quotient_map": lab_q, "h_good": h_good, "h_bad": h_bad, "good_witness_H": good_witness, "bad_failures": bad_failures, "checks": lab_checks},
    CHECK_DIR / "applied-lab-quotient-factorization-checks.json",
)

G_lab = nx.DiGraph()
for x in lab_X:
    G_lab.add_node(f"x{x}", layer="X", label=str(x))
for cls in lab_classes:
    G_lab.add_node(cls, layer="quotient", label=cls)
for value in sorted(set(h_good.values())):
    G_lab.add_node(f"z{value}", layer="Z", label=str(value))
for x in lab_X:
    G_lab.add_edge(f"x{x}", lab_q[x])
for cls, value in good_witness.items():
    G_lab.add_edge(cls, f"z{value}")
pos_lab = {}
for i, x in enumerate(lab_X):
    pos_lab[f"x{x}"] = (0, len(lab_X) - i)
for i, cls in enumerate(lab_classes):
    pos_lab[cls] = (1.35, len(lab_X) - 1.8 - i * 2.3)
for i, value in enumerate(sorted(set(h_good.values()))):
    pos_lab[f"z{value}"] = (2.7, len(lab_X) - 2.5 - i * 3.0)
fig, ax = plt.subplots(figsize=(10, 6.4))
node_colors = [{"X": "#dbeafe", "quotient": "#fed7aa", "Z": "#dcfce7"}[G_lab.nodes[node]["layer"]] for node in G_lab.nodes]
nx.draw_networkx_edges(G_lab, pos_lab, ax=ax, arrows=True, arrowstyle="-|>", arrowsize=13, width=1.1, edge_color="#64748b")
nx.draw_networkx_nodes(G_lab, pos_lab, ax=ax, node_size=900, node_color=node_colors, edgecolors="#334155")
nx.draw_networkx_labels(G_lab, pos_lab, labels={node: G_lab.nodes[node]["label"] for node in G_lab.nodes}, ax=ax, font_size=8)
ax.text(0, 0.4, "X", ha="center", fontsize=11, weight="bold")
ax.text(1.35, 0.4, "X/~", ha="center", fontsize=11, weight="bold")
ax.text(2.7, 0.4, "Z", ha="center", fontsize=11, weight="bold")
ax.text(1.35, -0.25, "Good rule descends; bad rule fails on " + bad_failures[0]["class"] + ".", ha="center", fontsize=9)
ax.set_title("Applied quotient lab: h factors exactly when it is constant on classes", fontsize=13, weight="bold")
ax.axis("off")
fig.tight_layout()
lab_fig_path = save_matplotlib(fig, FIG_DIR / "applied-lab-quotient-factorization.png")
plt.close(fig)

assert all(lab_checks.values())
display_artifact(rel_to_cwd(lab_fig_path), width=850)
{"good_witness_H": good_witness, "bad_failures": bad_failures}
```

## Takeaways

- A set is controlled by its membership criterion. Equality and subset relations are proved by checking arbitrary-element membership in both directions.
- Complements and set differences require a declared ambient set. In computation that ambient set is the universe `U`; in proofs it is usually supplied by context.
- Relations are subsets of products. Equivalence relations produce partitions and quotient sets; partial orders produce comparability structure, and maximal does not always mean maximum.
- Functions are relations with domain and codomain data. Images and preimages have different algebraic behavior: preimages preserve unions, intersections, and differences exactly, while images require care.
- Products record coordinated choices, indexed families generalize sequences, disjoint unions tag copies to keep them separate, and quotient maps collapse equivalence classes.
- Countability is the existence of a listing. Diagonal enumeration of `N x N` and reduced-fraction enumeration are the patterns behind many later countability arguments.
- The axiom of choice can be read as an abstract guarantee of simultaneous selections from nonempty indexed families. In finite code we choose explicitly; in general topology the theorem supplies the chooser.
