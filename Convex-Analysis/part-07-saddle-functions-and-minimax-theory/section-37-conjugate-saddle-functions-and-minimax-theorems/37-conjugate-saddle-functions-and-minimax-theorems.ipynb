## Translation Guide

Part VII adds the convex-concave viewpoint. The visuals distinguish descent in primal variables from ascent in dual variables, then use that split to explain saddle points and minimax equalities.

For this section, the translation into computational language is centered on saddle conjugate, minimax theorem, inverse bifunction, existence condition. The notebook treats those ideas as objects with coordinates, inequalities, sampled witnesses, and explicit residuals. That is deliberately narrower than the full theorem set of the printed section, but it is broad enough to make the main geometry visible and to support later refinement by a section worker.

- `saddle conjugate`: a conjugacy operation adapted to convex-concave functions.
- `minimax theorem`: a theorem guaranteeing equality of optimization orders.
- `inverse bifunction`: the bifunction operation related to reversing the pairing.
- `existence condition`: a hypothesis ensuring the saddle or minimax value is attained.
